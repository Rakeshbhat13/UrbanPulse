```python
from ultralytics import YOLO
import cv2
import pandas as pd

model = YOLO("yolo11n.pt")

video_path = "test_video.mp4"

cap = cv2.VideoCapture(video_path)

fps = cap.get(cv2.CAP_PROP_FPS)

data = []

frame_number = 0

while True:
    ret, frame = cap.read()

    if not ret:
        break

    results = model.track(
        frame,
        classes=[0],
        conf=0.3,
        tracker="bytetrack.yaml",
        persist=True,
        verbose=False
    )

    boxes = results[0].boxes

    if boxes.id is not None:

        ids = boxes.id.cpu().numpy()
        coords = boxes.xyxy.cpu().numpy()
        confidences = boxes.conf.cpu().numpy()

        for person_id, box, confidence in zip(
            ids, coords, confidences
        ):

            x1, y1, x2, y2 = box

            # Center of bounding box
            center_x = (x1 + x2) / 2
            center_y = (y1 + y2) / 2

            data.append({
                "frame": frame_number,
                "person_id": int(person_id),
                "center_x": center_x,
                "center_y": center_y,
                "confidence": confidence
            })

    frame_number += 1

cap.release()

df = pd.DataFrame(data)

print("Total detections:", len(df))
print("Unique people tracked:", df["person_id"].nunique())

df.head(10)
```

```python
from ultralytics import YOLO
import cv2
import pandas as pd
import numpy as np

model = YOLO("yolo11n.pt")

video_path = "umn_crowd.avi"

cap = cv2.VideoCapture(video_path)
fps = cap.get(cv2.CAP_PROP_FPS)

data = []
frame_number = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    results = model.track(
        frame,
        classes=[0],
        conf=0.3,
        tracker="bytetrack.yaml",
        persist=True,
        verbose=False
    )

    boxes = results[0].boxes

    if boxes.id is not None:

        ids = boxes.id.cpu().numpy()
        coords = boxes.xyxy.cpu().numpy()
        confidences = boxes.conf.cpu().numpy()

        for person_id, box, confidence in zip(
            ids, coords, confidences
        ):

            x1, y1, x2, y2 = box

            center_x = (x1 + x2) / 2
            center_y = (y1 + y2) / 2

            data.append({
                "frame": frame_number,
                "person_id": int(person_id),
                "center_x": center_x,
                "center_y": center_y,
                "confidence": float(confidence)
            })

    frame_number += 1

    if frame_number % 500 == 0:
        print("Processed:", frame_number, "frames")

cap.release()

df = pd.DataFrame(data)

print("\nProcessing complete!")
print("Total detections:", len(df))
print("Unique tracked people:", df["person_id"].nunique())
```

```python
df = df.sort_values(["person_id", "frame"]).copy()

# Movement between consecutive frames
df["dx"] = df.groupby("person_id")["center_x"].diff()
df["dy"] = df.groupby("person_id")["center_y"].diff()

# Time between frames
time_per_frame = 1 / fps

# Distance moved in pixels
df["distance"] = np.sqrt(
    df["dx"]**2 + df["dy"]**2
)

# Speed in pixels/second
df["speed"] = df["distance"] / time_per_frame

# Direction in degrees
df["direction"] = np.degrees(
    np.arctan2(df["dy"], df["dx"])
)

# First observation of each person has no previous position
df["speed"] = df["speed"].fillna(0)
df["direction"] = df["direction"].fillna(0)

print(df.head(15))
```

```python
# Create crowd-level features for every frame

crowd_features = df.groupby("frame").agg(
    person_count=("person_id", "nunique"),
    avg_speed=("speed", "mean"),
    median_speed=("speed", "median"),
    max_speed=("speed", "max"),
    speed_std=("speed", "std")
).reset_index()

# Replace missing standard deviation
crowd_features["speed_std"] = (
    crowd_features["speed_std"].fillna(0)
)

print(crowd_features.head(20))
```

```python
print("Total frames:", len(crowd_features))
print("Features:", crowd_features.columns.tolist())
```

```python
# Calculate crowd direction consistency

direction_features = df.groupby("frame").agg(
    avg_direction=("direction", "mean"),
    direction_std=("direction", "std")
).reset_index()

direction_features["direction_std"] = (
    direction_features["direction_std"].fillna(0)
)

# Combine with the previous crowd features
crowd_features = crowd_features.merge(
    direction_features,
    on="frame",
    how="left"
)

print(crowd_features.head(20))
```

```python
import cv2

video_path = "umn_crowd.avi"
cap = cv2.VideoCapture(video_path)

marked_frames = []
frame_number = 0

while True:
    ok, frame = cap.read()
    if not ok:
        break

    height, width = frame.shape[:2]

    # The red text is in the upper-left area.
    roi = frame[:int(height * 0.18), :int(width * 0.50)]
    hsv = cv2.cvtColor(roi, cv2.COLOR_BGR2HSV)

    # Red spans both ends of the HSV hue range.
    red_low = cv2.inRange(hsv, (0, 100, 100), (10, 255, 255))
    red_high = cv2.inRange(hsv, (170, 100, 100), (180, 255, 255))
    red_pixels = cv2.countNonZero(red_low | red_high)

    marked_frames.append(red_pixels > 25)
    frame_number += 1

cap.release()

# Convert consecutive marked frames into intervals; bridge gaps of up to 2 frames.
intervals = []
start = None
last_marked = None

for i, marked in enumerate(marked_frames):
    if marked:
        if start is None:
            start = i
        last_marked = i
    elif start is not None and i - last_marked > 2:
        intervals.append((start, last_marked))
        start = None
        last_marked = None

if start is not None:
    intervals.append((start, last_marked))

print("Frames scanned:", len(marked_frames))
print("Red-overlay intervals:")
for start, end in intervals:
    print(f"{start}–{end}  ({start / 30:.1f}s–{end / 30:.1f}s)")
```

```python
import cv2
import numpy as np

video_path = "umn_crowd.avi"
cap = cv2.VideoCapture(video_path)

marked_frames = []
frame_number = 0

while True:
    ok, frame = cap.read()
    if not ok:
        break

    height, width = frame.shape[:2]
    roi = frame[:int(height * 0.18), :int(width * 0.65)]
    hsv = cv2.cvtColor(roi, cv2.COLOR_BGR2HSV)

    red = (
        cv2.inRange(hsv, (0, 100, 100), (10, 255, 255))
        | cv2.inRange(hsv, (170, 100, 100), (180, 255, 255))
    )

    ys, xs = np.where(red > 0)

    # The overlay is a wide, short line of text.
    is_overlay = (
        len(xs) >= 25
        and (xs.max() - xs.min()) >= 60
        and (ys.max() - ys.min()) <= 20
    )

    marked_frames.append(is_overlay)
    frame_number += 1

cap.release()

# Merge tiny gaps in the detected text.
intervals = []
start = None
last_marked = None

for i, marked in enumerate(marked_frames):
    if marked:
        if start is None:
            start = i
        last_marked = i
    elif start is not None and i - last_marked > 2:
        intervals.append((start, last_marked))
        start = None
        last_marked = None

if start is not None:
    intervals.append((start, last_marked))

print("Frames scanned:", len(marked_frames))
print("Overlay intervals:")
for start, end in intervals:
    print(f"{start}–{end}")
```

```python
import numpy as np
from sklearn.preprocessing import StandardScaler

total_frames = len(crowd_features)

# Approximate starts of the 11 clips, based on the contact sheets.
clip_starts = [0, 625, 1453, 1986, 2685, 3429, 4035, 4929, 5597, 6235, 6913]

# Assign each frame to a clip.
clip_ids = np.zeros(total_frames, dtype=int)

for i, start in enumerate(clip_starts):
    end = clip_starts[i + 1] if i + 1 < len(clip_starts) else total_frames
    clip_ids[start:end] = i + 1

feature_cols = [
    "person_count",
    "avg_speed",
    "median_speed",
    "max_speed",
    "speed_std",
    "avg_direction",
    "direction_std",
]

# Hold out one clip from each scene for testing.
test_clip_ids = [2, 6, 10]
train_clip_ids = [i for i in range(1, 12) if i not in test_clip_ids]

# Fit normalization only on training clips.
train_rows = np.isin(clip_ids, train_clip_ids)
scaler = StandardScaler()
scaler.fit(crowd_features.loc[train_rows, feature_cols])

X_scaled = scaler.transform(crowd_features[feature_cols]).astype(np.float32)
labels = crowd_features["label"].to_numpy(dtype=np.int64)

# 30 frames = 1 second; stride 5 reduces highly overlapping windows.
sequence_length = 30
stride = 5

def make_sequences(selected_clip_ids):
    X_sequences, y_sequences = [], []

    for clip_id in selected_clip_ids:
        start = clip_starts[clip_id - 1]
        end = clip_starts[clip_id] if clip_id < len(clip_starts) else total_frames

        for window_start in range(start, end - sequence_length + 1, stride):
            window_end = window_start + sequence_length
            X_sequences.append(X_scaled[window_start:window_end])
            # Predict whether the final frame in the window is abnormal.
            y_sequences.append(labels[window_end - 1])

    return np.asarray(X_sequences), np.asarray(y_sequences)

X_train, y_train = make_sequences(train_clip_ids)
X_test, y_test = make_sequences(test_clip_ids)

print("Train:", X_train.shape, y_train.shape)
print("Test: ", X_test.shape, y_test.shape)
print("Train label counts [normal, abnormal]:", np.bincount(y_train, minlength=2))
print("Test label counts  [normal, abnormal]:", np.bincount(y_test, minlength=2))
```

```python
import numpy as np
from sklearn.preprocessing import StandardScaler

train_clip_ids = [1, 3, 5, 7, 8, 11]
val_clip_ids   = [4, 9]
test_clip_ids  = [2, 6, 10]

# Fit feature scaling on training clips only.
train_rows = np.isin(clip_ids, train_clip_ids)
scaler = StandardScaler()
scaler.fit(crowd_features.loc[train_rows, feature_cols])

X_scaled = scaler.transform(crowd_features[feature_cols]).astype(np.float32)
labels = crowd_features["label"].to_numpy(dtype=np.int64)

def make_sequences(selected_clip_ids):
    X_sequences, y_sequences = [], []

    for clip_id in selected_clip_ids:
        start = clip_starts[clip_id - 1]
        end = clip_starts[clip_id] if clip_id < len(clip_starts) else total_frames

        for window_start in range(start, end - sequence_length + 1, stride):
            window_end = window_start + sequence_length
            X_sequences.append(X_scaled[window_start:window_end])
            y_sequences.append(labels[window_end - 1])

    return np.asarray(X_sequences), np.asarray(y_sequences)

X_train, y_train = make_sequences(train_clip_ids)
X_val, y_val = make_sequences(val_clip_ids)
X_test, y_test = make_sequences(test_clip_ids)

print("Train:", X_train.shape, "labels [normal, abnormal]:",
      np.bincount(y_train, minlength=2))
print("Validation:", X_val.shape, "labels [normal, abnormal]:",
      np.bincount(y_val, minlength=2))
print("Test:", X_test.shape, "labels [normal, abnormal]:",
      np.bincount(y_test, minlength=2))
```

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.metrics import AUC, Precision, Recall

tf.keras.utils.set_random_seed(42)

counts = np.bincount(y_train, minlength=2)
class_weight = {
    0: len(y_train) / (2 * counts[0]),
    1: len(y_train) / (2 * counts[1]),
}
print("Class weights:", class_weight)

model = Sequential([
    Input(shape=(sequence_length, len(feature_cols))),
    LSTM(32, dropout=0.2),
    Dense(16, activation="relu"),
    Dropout(0.2),
    Dense(1, activation="sigmoid"),
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        Precision(name="precision"),
        Recall(name="recall"),
        AUC(name="roc_auc"),
        AUC(curve="PR", name="pr_auc"),
    ],
)

early_stopping = EarlyStopping(
    monitor="val_pr_auc",
    mode="max",
    patience=8,
    restore_best_weights=True,
)

history = model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=50,
    batch_size=32,
    class_weight=class_weight,
    callbacks=[early_stopping],
    verbose=1,
)

model.save("urbanpulse_lstm.keras")
print("Saved: urbanpulse_lstm.keras")
print("Best validation PR-AUC:", max(history.history["val_pr_auc"]))
```

```python
import numpy as np
from sklearn.metrics import (
    precision_recall_curve,
    f1_score,
    classification_report,
    confusion_matrix,
    average_precision_score,
    roc_auc_score,
)

# Choose the threshold using validation data only.
val_probs = model.predict(X_val, verbose=0).ravel()
precision, recall, thresholds = precision_recall_curve(y_val, val_probs)

f1_values = (
    2 * precision[:-1] * recall[:-1]
    / (precision[:-1] + recall[:-1] + 1e-8)
)
best_index = np.argmax(f1_values)
best_threshold = thresholds[best_index]

print(f"Validation threshold: {best_threshold:.3f}")
print(f"Validation precision: {precision[best_index]:.3f}")
print(f"Validation recall: {recall[best_index]:.3f}")
print(f"Validation F1: {f1_values[best_index]:.3f}")

# Evaluate the held-out clips using that threshold.
test_probs = model.predict(X_test, verbose=0).ravel()
test_preds = (test_probs >= best_threshold).astype(int)

print("\nTest confusion matrix [ [TN, FP], [FN, TP] ]:")
print(confusion_matrix(y_test, test_preds, labels=[0, 1]))

print("\nTest classification report:")
print(classification_report(
    y_test,
    test_preds,
    target_names=["Normal", "Abnormal"],
    zero_division=0,
))

print("Test PR-AUC:", round(average_precision_score(y_test, test_probs), 3))
print("Test ROC-AUC:", round(roc_auc_score(y_test, test_probs), 3))
```

```python
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report

results = test_results.sort_values(["clip_id", "frame"]).copy()
results["alert"] = 0
alerts = []

for clip_id in test_clip_ids:
    indices = results.index[results["clip_id"] == clip_id].tolist()

    active = False
    positive_run = 0
    negative_run = 0
    alert_start = None

    for row_index in indices:
        frame = int(results.at[row_index, "frame"])
        is_positive = results.at[row_index, "probability"] >= best_threshold

        if not active:
            positive_run = positive_run + 1 if is_positive else 0

            if positive_run >= 3:
                active = True
                alert_start = frame
                negative_run = 0
        else:
            negative_run = 0 if is_positive else negative_run + 1

            if negative_run >= 3:
                active = False
                alerts.append((clip_id, alert_start, frame))
                alert_start = None
                positive_run = 0
                negative_run = 0

        results.at[row_index, "alert"] = int(active)

    if active:
        alerts.append((clip_id, alert_start, int(results.at[indices[-1], "frame"])))

print("Debounced alert confusion matrix [ [TN, FP], [FN, TP] ]:")
print(confusion_matrix(results["actual"], results["alert"], labels=[0, 1]))

print(classification_report(
    results["actual"],
    results["alert"],
    target_names=["Normal", "Abnormal"],
    zero_division=0,
))

print("Alert intervals (clip, start frame, end frame):")
for alert in alerts:
    print(alert)
```
